## 1.3 Pandas 자료구조

분석을 위해 다양한 소스(source)로부터 수집하는 데이터는 형태나 속성이 매우 다양하다. 특히, 서로 다른 형식을 갖는 여러 종류의 데이터를 컴퓨터가 이해할 수 있도록 동일한 형식을 갖는 구조로 통합할 필요가 있다. 판다스의 일차적인 목적은 **형식적으로 서로 다른 여러 가지 유형의 데이터를 공통의 포맷으로 정리**하는 것이다.  

판다스는 **시리즈(Series)와 데이터프레임(DataFrame)이라는 구조화된 데이터 형식**을 제공한다. 서로 다른 종류의 데이터를 한곳에 담는 그릇(컨테이너)이 된다. 다만, 시리즈는 1차원 배열이고, 데이터프레임이 2차원 배열이라는 점에서 차이가 있다. 특히, 행과 열로 이루어진 2차원 구조의 데이터프레임은 데이터 분석 실무에서 자주 사용된다.

### 1.3.1 Series

Series는 **1차원 배열** 형태의 자료구조입니이다. NumPy의 1차원 배열과 유사하지만, 가장 큰 차이점은 데이터의 위치를 가리키는 **인덱스(Index) 라벨이 데이터와 1:1로 결합**되어 있다는 점이다. 

- **구조**: `인덱스(라벨) + 값(Value)`
- **특징**: 한 가지 데이터 타입(dtype)만 담을 수 있다. 별도의 인덱스를 지정하지 않으면 0부터 시작하는 정수형 인덱스가 자동으로 생성된다.

```python
# 원소가 없는 빈 Series 객체를 만들기

pd.Series() # Series([], dtype: object)
```

```python
# 배열이 아닌 스칼라(Scalar) 상수 값으로 Series 객체를 만들기
print(pd.Series(5))

print(pd.Series(5, index=["a", "b", "c"]))
```

```python
# 리스스트를 사용한 Series 생성
import pandas as pd

# 리스트를 사용한 시리즈 생성 (인덱스 직접 지정)
population = pd.Series([940, 330, 290], index = ['서울', '부산', '인천'])
print(population)

# index 확인하기
print(population.index) # Index(['서울', '부산', '인천'], dtype='str')

# 배열 확인하기
print(population.values) # [940 330 290]
```

```python
# 딕셔너리를 시리즈(Series)로 변환
# 딕셔너리의 키(k)는 시리즈의 인덱스에 대응하고, 딕셔너리의 각 키에 매칭되는 값(v)이 시리즈의 데이터 값(원소)로 변환


# k:v 구조를 갖는 딕셔너리를 만들고, 변수 dict_data에 저장
dict_data = {'a': 1, 'b': 2, 'c': 3}
print(dict_data)            # {'a': 1, 'b': 2, 'c': 3}

# 판다스 Series() 함수로 딕셔너리(dict_data)를 시리즈로 변환. 변수 sr에 저장
sr = pd.Series(dict_data)

# 변수 sr의 자료형 출력
print(type(sr))            # <class 'pandas.core.series.Series'>

# 주피터 환경에서 변수 출력
print(sr)

# 인덱스 배열은 변수 idx에 저장
idx = sr.index
print(idx)

# 데이터 값 배열은 변수 val에 저장
val = sr.values
print(val)
```

```python
# 파이썬 리스트(List) 객체를 판다스 시리즈(Series)로 변환

import pandas as pd

# 리스트를 시리즈로 변환하여 변수 sr에 저장
list_data = ['2019-01-02', 3.14, 'ABC', 100, True]
sr = pd.Series(list_data)
print(sr)

# 인덱스 배열은 변수 idx에 저장
idx = sr.index
print(idx)                         # RangeIndex(start=0, stop=5, step=1)

# 데이터 값 배열은 변수 val에 저장
val = sr.values
print(val)                         # ['2019-01-02' 3.14 'ABC' 100 True]
```

```python
# 파이썬 투플(Tuple) 객체를 판다스 시리즈(Series)로 변환

import pandas as pd

# 투플을 시리즈로 변환(index 옵션에 인덱스 이름을 지정)
tup_data = ('영인', '2010-05-01', '여', True)
sr = pd.Series(tup_data, index=['이름', '생년월일', '성별', '학생여부'])
print(sr)

'''
이름              영인
생년월일    2010-05-01
성별               여
학생여부          True
dtype: object
'''

# 인덱스 배열은 변수 idx에 저장
idx = sr.index
print(idx)           # Index(['이름', '생년월일', '성별', '학생여부'], dtype='str')

# 데이터 값 배열은 변수 val에 저장
val = sr.values
print(val)            # ['영인' '2010-05-01' '여' True]


sr = pd.Series(tup_data)
# 인덱스 배열은 변수 idx에 저장
idx = sr.index
print(idx)              # RangeIndex(start=0, stop=4, step=1

# 데이터 값 배열은 변수 val에 저장
val = sr.values
print(val)              # ['영인' '2010-05-01' '여' True]
```

### DataFrame

 DataFrame은 행(Row)과 열(Column)로 구성된 **2차원 표 형태**의 자료구조다. 데이터 분석에서 가장 많이 사용하는 객체로, 쉽게 말해 **여러 개의 Series가 열(Column)로 묶여 있는 구조**다. 
• **구조**: `행 인덱스(Row Index) + 열 이름(Column Name) + 값(Values)`
• **특징**: 표의 가로축은 행 인덱스(row index), 세로축은 열 이름(column name 또는 column label)으로 접근한다. 각 열(Column)마다 서로 다른 데이터 타입(문자열, 숫자, 날짜 등)을 가질 수 있다.  데이터프레임의 각 열은 공통의 속성을 갖는 일련의 데이터를 나타낸다.  각 행은 개별 관측대상에 대한 다양한 속성 데이터들의 모음인 레코드(record) 낸다.

**파이썬 딕셔너리(Dictionary) 객체를 판다스 데이터프레임(DataFrame)으로 변환**

 ① 같은 길이(원소의 개수가 동일한)의 배열 여러 개가 필요. 데이터프레임은 여러 개의 시리즈(열, column)를 모아 놓은 집합.

② 판다스 DataFrame() 함수를 사용. 여러 개의 리스트를 원소로 갖는 딕셔너리를 함수에 전달하는 방식을 주로 활용.

③ 딕셔너리의 값(v)에 해당하는 각 리스트가 시리즈로 변환되어 데이터프레임의 각 열이 된다.

④딕셔너리의 키(k)는 각 시리즈의 이름으로 변환되어, 최종적으로 데이터프레임의 열 이름이 된다.

```python
import pandas as pd

# 열이름을 key로 하고, 리스트를 value로 갖는 딕셔너리 정의(2차원 배열)
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 판다스 DataFrame() 함수로 딕셔너리를 데이터프레임으로 변환. 변수 df에 저장.
df = pd.DataFrame(dict_data)

# df의 자료형 출력
print(type(df))


# 변수 df에 저장되어 있는 데이터프레임 객체를 출력
print(df)
```

```python
# 딕셔너리를 사용한 데이터프레임 생성
data = {
    '나이':[19,30,40],
    '직업': ['학생', '회사원', '연구원']
}

df = pd.DataFrame(data, index=['철수', '영희', '민수'])
print(df)


'''
    나이   직업
철수  19   학생
영희  30  회사원
민수  40  연구원
'''

#행 인덱스 확인하기
print(df.index)      

#열 이름 확인하기
print(df.columns) 
```

```python
import pandas as pd

# 행 인덱스/열 이름 지정하여, 데이터프레임 만들기
df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']],
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

# 데이터프레임 출력하기
print(df)    

#행 인덱스 확인하기
print(df.index)      

#열 이름 확인하기
print(df.columns)  


# 행 인덱스, 열 이름 변경하기
df.index=['학생1', '학생2']
df.columns=['연령', '남녀', '소속']

print(df)           
```

```python
import pandas as pd

# 행 인덱스/열 이름 지정하여, 데이터프레임 만들기
df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']],
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

# 행 인덱스, 열 이름 확인하기
print(df)

# 열 이름 중, '나이'를 '연령'으로, '성별'을 '남녀'로, '학교'를 '소속'으로 바꾸기
df = df.rename(columns={'나이':'연령', '성별':'남녀', '학교':'소속'})

# df 출력(변경 후)
print(df)

# df의 행 인덱스 중에서, '준서'를 '학생1'로, '예은'을 '학생2'로 바꾸기
df = df.rename(index={'준서':'학생1', '예은':'학생2' })

# df 출력(변경 후)
print(df)
```

### Index

Index는 Series와 DataFrame의 **행(또는 열)의 이름을 저장하는 메타데이터 객체이**다. 데이터의 주소 역할을 하며, 데이터를 식별하고 정렬할 때 기준이 된다.
• **특징**:
    ◦  인덱스는 자기와 짝을 이루는 **원소의 순서와 주소를 저장**   
    ◦ 인덱스를 잘 활용하면 데이터 값의 탐색, 정렬, 선택, 결합 등 데이터 조작을 쉽게 할 수 있다. 
    ◦     **인덱스의 종류 (2가지)**   

① 정수형 위치 인덱스(integer position)

② 인덱스 이름(index name) 또는 인덱스 라벨(index label)

## NumPy와 Pandas 비교

Pandas는 내부적으로 NumPy를 기반으로 구축되었지만, 제공하는 기능과 목적이 다르다.

### 1. 4.1 배열(ndarray)과  데이터프레임(DataFrame) 차이

| **구분** | **NumPy (ndarray)** | **Pandas (DataFrame / Series)** |
| --- | --- | --- |
| **차원 구조** | $N$차원 배열 (`ndarray`) | 1차원(`Series`), 2차원 표 형식(`DataFrame`) |
| **데이터 타입** | 단일 타입만 허용 (Homogeneous) | 열(Column)별 서로 다른 타입 허용 (Heterogeneous) |
| **인덱싱** | 암묵적 위치 인덱스 (0, 1, 2...) | 명시적 행/열 라벨 (문자열, 날짜 등 인덱스 지정 가능) |
| **데이터 형태** | 순수 수치/행렬 구조 | 엑셀/SQL 테이블과 유사한 표 구조 |

### 1.4.2 활용 목적 비교

- **NumPy:**
    - **고속 수치 계산 및 선형대수 연산**에 최적화되어 있다.
    - 머신러닝/딥러닝 모델 내부의 가중치 계산, 이미지/음성 데이터 처리, 복잡한 수학적 알고리즘 구현에 주로 사용한다.
- **Pandas:**
    - **실세상(Real-world) 데이터의 정제, 전처리 및 분석**에 최적화되어 있다.
    - CSV, Excel, SQL 데이터베이스 등에서 표 데이터를 불러와 결측치 처리, 그룹화, 필터링, 병합 작업을 수행할 때  사용한다.

### 장단점 비교

#### **NumPy**

- **장점:**
    - **빠른 속도와 적은 메모리:** C언어 기반 구현 및 동일 타입 메모리 배치로 대용량 수치 계산 시 연산 속도가 매우 빠릅니다.
    - **고차원 지원:** 3차원 이상의 다차원 데이터 다루기에 유리하다.
- **단점:**
    - **라벨 부재:** 열 이름을 지정할 수 없어 위치로만 접근하므로 데이터 탐색 시 직관성이 떨어집니다.
    - **혼합 타입 불가:** 정수, 실수, 문자열이 섞인 데이터를 다루기 어렵다.

#### **Pandas**

- **장점:**
    - **높은 편의성:** 컬럼명 기반 접근, 결측치 자동 처리, 시계열 데이터 관리 등 강력한 전처리 API를 제공한다.
    - **입출력 및 가공 지원:** 파일 읽기/쓰기, 데이터 합치기(`merge`, `join`), 그룹별 집계(`groupby`)가 매우 간편하다.
- **단점:**
    - **높은 메모리 사용량:** 인덱스와 메타데이터를 함께 저장하므로 순수 NumPy 배열에 비해 메모리 점유율이 높다.
    - **연산 속도:** 순수 수치 연산만 수행할 때는 NumPy보다 다소 느릴 수 있다.

## CSV 파일 읽기

데이터 값을 쉼표(,)로 구분하고 있다는 의미로 CSV(comma-separated values)라고 부르는 텍스트 파일이다. 쉼표(,)로 열을 구분하고 줄바꿈으로 행을 구분한다.
read_csv() 함수에 확장자(.csv)를 포함하여 파일경로(파일명)을 입력하면, CSV 파일을 읽어와서 데이터프레임으로 변환한다.

read_csv()

csv 파일 -> 데이터 프레임: paadas.read_csv("파일 경로(이름)")

### 인코딩 지정

- 한글이 깨지는 경우 문서 형식에 따라 인코딩을 지정합니다.
- `encoding='utf-8'` (기본 표준)
- `encoding='cp949'` 또는 `encoding='euc-kr'` (한글 윈도우 환경에서 생성된 파일)

### 구분자 지정

- 쉼표(`,`)가 아닌 탭(`\t`), 세미콜론(`;`) 등으로 구분된 경우 지정합니다.
- `sep='\t'` (TSV 파일 읽기)

### 헤더 설정

- 첫 행을 컬럼명으로 사용할지 지정합니다.
- `header=0`: 첫 번째 행을 컬럼명으로 사용 (기본값)
- `header=None`: 데이터에 헤더가 없을 때 사용하며, 0, 1, 2... 식의 숫자 인덱스가 부여됩니다.
- `names=['A', 'B', 'C']`: 헤더가 없거나 변경하고 싶을 때 직접 컬럼명을 부여할 수 있습니다.
