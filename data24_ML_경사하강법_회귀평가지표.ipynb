### 경사 하강법(gradient_descent_steps) _MSE(평균제곱오차계산식)
- 경사하강법(Gradient Descent)은 함수의 손실(오차)을 최소화하기 위해 기울기(Gradient)가 낮은 방향으로 매개변수를 반복해서 업데이트하는 최적화 알고리즘
- 파라미터 (Parameter) :  스스로 학습하며 최적의 값을 찾아가는 변수 -w, b

- 하이퍼파라미터 (Hyperparameter) : 모델이 학습을 시작하기 전에, 사람(개발자)이 외부에서 직접 설정해 주는 값 - learning_rate, l1, l2

```python
"""
============================================================================
경사하강법으로 선형 회귀 직접 구현하기 — 전체 주석판
============================================================================

[이 코드가 하는 일]
    y = wX + b 라는 직선을 그어 놓고,
    "실제 정답과 얼마나 틀렸는지(손실)"를 재고,
    그 손실이 줄어드는 방향으로 w와 b를 조금씩 고치는 일을 반복한다.

[핵심 수식 3줄]
    예측    ŷ = wX + b
    손실    J(w,b) = (1/N) Σ (y - ŷ)²          <- MSE, 평균제곱오차
    갱신    w := w - α · ∂J/∂w                 <- 경사하강법
============================================================================
"""

import numpy as np
import matplotlib.pyplot as plt


# ===========================================================================
# 1. 한 스텝만큼의 업데이트량을 계산하는 함수
# ===========================================================================
def get_weight_updates(w, b, X, y, learning_rate=0.01):
    """
    현재 w, b에서 '다음 스텝에 얼마나 움직여야 하는지'를 계산해서 돌려준다.

    파라미터
    --------
    w : (1, 1) 배열   현재 가중치(기울기)
    b : (1, 1) 배열   현재 편향(y절편)
    X : (N, 1) 배열   입력 데이터
    y : (N, 1) 배열   실제 정답
    learning_rate : float   한 걸음의 보폭(α). 하이퍼파라미터.

    반환
    ----
    w_update, b_update : 각각 w와 b에 '더해줄' 값
    """
    N = len(y)                      # 데이터 개수. 여기서는 100

    # -----------------------------------------------------------------
    # [1단계] 현재 직선으로 예측해보고, 얼마나 틀렸는지 잰다
    # -----------------------------------------------------------------
    # np.dot(X, w.T) : (N,1) @ (1,1) = (N,1)
    #   X가 (100,1)이고 w가 (1,1)이므로 w.T도 (1,1).
    #   행렬곱 결과는 (100,1) — 데이터 100개 각각의 예측값.
    #   피처가 여러 개(예: (N,3))로 늘어나도 w를 (1,3)으로만 바꾸면
    #   이 코드가 그대로 동작하도록 dot 연산으로 짠 것이다.
    # + b : b는 (1,1)이지만 넘파이 브로드캐스팅으로 100행 전부에 더해진다.
    y_pred = np.dot(X, w.T) + b     # 예측값 ŷ, shape (N, 1)

    # diff = 실제값 - 예측값  (순서 주의! 아래 부호 규약의 근거가 된다)
    #   diff > 0  ->  과소예측 (실제보다 낮게 봤다) -> 직선을 위로 올려야 함
    #   diff < 0  ->  과대예측 (실제보다 높게 봤다) -> 직선을 내려야 함
    diff = y - y_pred               # 오차 e, shape (N, 1)

    # -----------------------------------------------------------------
    # [2단계] b의 업데이트도 dot 연산으로 처리하기 위한 1 벡터
    # -----------------------------------------------------------------
    # b의 기울기는 Σ(y - ŷ) 즉 '오차를 그냥 다 더한 것'이다.
    # np.sum(diff)로 해도 되지만, w와 똑같은 dot 형태로 맞추려고
    # 전부 1인 (N,1) 벡터를 만들어 내적한다.
    #   np.dot(b_factors.T, diff) = (1,N) @ (N,1) = (1,1) = Σ diff
    # 즉 "1을 곱해서 더한다 = 그냥 더한다"는 트릭이다.
    b_factors = np.ones((N, 1))

    # -----------------------------------------------------------------
    # [3단계] 기울기(gradient) 계산  ★ 여기가 2/N이 나오는 자리 ★
    # -----------------------------------------------------------------
    #
    #   손실 J = (1/N) Σ (y - ŷ)²  를 w로 미분하면
    #
    #       ∂J/∂w = -(2/N) Σ x·(y - ŷ)          ... (A)
    #       ∂J/∂b = -(2/N) Σ   (y - ŷ)          ... (B)
    #
    #   ▸ 계수 '2'는 어디서 왔나?
    #       제곱을 미분할 때 지수 2가 앞으로 내려온 것.
    #       d/dw (y - wx - b)² = 2(y - wx - b)·(-x)
    #
    #   ▸ '1/N'은 어디서 왔나?
    #       MSE가 오차 제곱의 '합'이 아니라 '평균'이기 때문.
    #       데이터가 100개든 10만 개든 기울기 크기를 비슷하게 유지해준다.
    #       (1/N이 없으면 데이터가 많아질수록 기울기가 커져서
    #        같은 학습률로는 발산한다. 실험으로 확인됨.)
    #
    #   ▸ 사실 2/N은 학습률에 흡수되는 상수다.
    #       빼도 되지만, 그러면 데이터 크기가 바뀔 때마다
    #       학습률을 다시 튜닝해야 한다. 그래서 붙여 둔다.
    #       (교재가 J = (1/2N)Σ 로 정의하는 것도 같은 이유.
    #        미분의 2와 1/2이 상쇄돼 계수가 1/N으로 깔끔해진다.)
    #
    #   ▸ 부호 규약 — 이 코드에서 가장 헷갈리는 부분 ★
    #       (A)식의 진짜 기울기에는 마이너스가 붙어 있다.
    #       그런데 아래 코드는 diff = (y - ŷ)를 그대로 쓰므로
    #       계산된 값은 '기울기'가 아니라 '기울기의 반대 부호',
    #       즉 -∂J/∂w 다.
    #
    #           w_update = +(2/N)·α·Σ x(y-ŷ) = -α · ∂J/∂w
    #
    #       경사하강법 공식이 w := w - α·∂J/∂w 이므로,
    #       이미 부호가 뒤집힌 w_update는 '더해야' 맞다.
    #       ->  w = w + w_update   (O)
    #       만약 diff를 (y_pred - y)로 짰다면 빼야 한다.
    #
    # np.dot(X.T, diff) : (1,N) @ (N,1) = (1,1) = Σ x·(y - ŷ)
    w_update = (2 / N) * learning_rate * np.dot(X.T, diff)

    # np.dot(b_factors.T, diff) : (1,N) @ (N,1) = (1,1) = Σ (y - ŷ)
    b_update = (2 / N) * learning_rate * np.dot(b_factors.T, diff)

    return w_update, b_update


# ===========================================================================
# 2. 위 함수를 epochs번 반복 호출하는 학습 루프
# ===========================================================================
def gradient_descent_steps(X, y, epochs=1000, learning_rate=0.1):
    """
    w, b를 0에서 출발시켜 epochs번 갱신한다.

    이 루프가 곧 '학습'이다. 신경망 학습 4단계 중
      ② 손실 계산 -> ③ 기울기 산출 -> ④ 매개변수 갱신
    을 반복하는 부분에 해당한다.
    (①미니배치 추출은 여기선 생략 — 매번 전체 데이터를 쓰는 '배치 경사하강법')
    """
    # 초기값을 0으로 둔다.
    # 선형 회귀는 손실면이 아래로 볼록한 그릇 모양(convex)이라
    # 어디서 출발하든 같은 최저점에 도착한다. 그래서 0으로 시작해도 안전하다.
    # (신경망은 convex가 아니라서 0 초기화를 하면 안 되고 무작위로 준다.)
    w = np.zeros((1, 1))
    b = np.zeros((1, 1))

    for ind in range(epochs):
        # 이번 스텝에 움직일 양을 계산
        w_update, b_update = get_weight_updates(
            w, b, X, y, learning_rate=learning_rate
        )

        # 갱신. 위에서 설명한 부호 규약 때문에 '+'가 맞다.
        # (w_update 안에 이미 마이너스 기울기가 들어 있음)
        w = w + w_update
        b = b + b_update

    return w, b


# ===========================================================================
# 3. 성능 평가 — MSE (평균제곱오차)
# ===========================================================================
def get_cost(y, y_pred):
    """
    MSE = (1/N) Σ (y - ŷ)²

    학습에 쓰는 손실 함수와 같은 식이다.
    다만 여기서는 '얼마나 잘 맞췄나 보고용'으로 쓴다.
    실제 보고할 때는 √를 씌운 RMSE가 더 낫다.
    단위가 원래 y와 같아져서 해석이 되기 때문.
    """
    N = len(y)
    cost = np.sum(np.square(y - y_pred)) / N
    return cost


# ===========================================================================
# 4. 데이터 생성 및 실행
# ===========================================================================
np.random.seed(42)          # 난수 고정 — 실행할 때마다 같은 결과가 나오게

# X: 0~2 사이 균등분포 난수 100개, shape (100, 1)
X = 2 * np.random.rand(100, 1)

# y: 정답 직선 y = 4X + 6 에 표준정규분포 노이즈를 얹은 것
#    np.random.randn(100,1)이 노이즈. 현실 데이터의 '흔들림'을 흉내낸 것.
y = 6 + 4 * X + np.random.randn(100, 1)

# 학습 실행
w, b = gradient_descent_steps(X, y, epochs=1000, learning_rate=0.05)

# -----------------------------------------------------------------
# 결과 출력
# -----------------------------------------------------------------
# ★ 중요 ★ 결과가 정확히 4와 6이 안 나오는 게 '정상'이다.
#    노이즈가 섞인 이 100개 샘플에 대한 진짜 최적해는
#    정규방정식으로 풀면 w=3.7701, b=6.2151 이고,
#    경사하강법이 바로 이 값에 도달한다.
#    즉 코드가 틀린 게 아니라 '주어진 데이터에 대한 정답'을 맞춘 것이다.
print(f'추정된 w: {w[0, 0]:.4f}  (데이터 생성에 쓴 값: 4, 최적해: 3.7701)')
print(f'추정된 b: {b[0, 0]:.4f}  (데이터 생성에 쓴 값: 6, 최적해: 6.2151)')

# 최종 예측값
# w[0,0]으로 스칼라를 꺼내 곱했다. b는 (1,1)이지만 브로드캐스팅으로 더해진다.
y_pred = w[0, 0] * X + b

print(f'Gradient Descent Total Cost (MSE): {get_cost(y, y_pred):.4f}')
print(f'RMSE: {np.sqrt(get_cost(y, y_pred)):.4f}  <- 이게 실제 오차 크기(단위 = y와 동일)')


# ===========================================================================
# 5. 시각화
# ===========================================================================
plt.figure(figsize=(8, 5))

# 실제 데이터 산점도
plt.scatter(X, y, label='Actual Data (y)', color='blue', alpha=0.6)

# 예측 직선
# ★ 주의 ★ plt.plot은 점들을 '배열 순서대로' 이어 그린다.
#    X가 무작위 순서라 정렬 없이 그리면 선이 앞뒤로 왔다갔다 하며 그려진다.
#    (지금은 모든 점이 같은 직선 위에 있어 겹쳐 보이지만,
#     곡선 모델이었다면 그래프가 엉망이 된다. 정렬하는 습관을 들이자.)
order = np.argsort(X.flatten())          # X를 오름차순으로 만드는 인덱스
plt.plot(X[order], y_pred[order],
         label='Fitted Line (y_pred)', color='red', linewidth=2)

plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression with Gradient Descent')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ===========================================================================
# [부록] 원본 코드에서 손봐야 할 점
# ===========================================================================
#
# 1) l1, l2 파라미터가 선언만 되고 안 쓰였다.
#    규제(regularization)를 구현하려면 기울기에 이렇게 더해야 한다.
#
#        # L2 (Ridge): 가중치를 0 쪽으로 서서히 끌어당김
#        w_update -= learning_rate * l2 * w
#
#        # L1 (Lasso): 가중치를 아예 0으로 만들어 변수 선택 효과
#        w_update -= learning_rate * l1 * np.sign(w)
#
#    (부호가 -인 이유: 규제항 자체도 손실이므로 줄이는 방향으로 가야 함.
#     편향 b에는 보통 규제를 걸지 않는다.)
#    쓰지 않을 거라면 함수 시그니처에서 지우는 게 낫다.
#
# 2) 학습 곡선을 남기면 디버깅이 훨씬 쉽다.
#    루프 안에서 100회마다 get_cost를 기록해 그려보면
#    "수렴했는지 / 발산했는지 / 학습률이 적절한지"가 한눈에 보인다.
#
# 3) 학습률 감각 (이 데이터 기준, 1000 epoch)
#        0.001  ->  w=1.37  아직 한참 멀었음 (너무 느림)
#        0.05   ->  w=3.77  정확히 수렴      (적절)
#        0.35   ->  발산                     (너무 큼)
#
# 4) 이 코드는 매번 전체 데이터를 쓰는 '배치 경사하강법'이다.
#    데이터가 커지면 매 스텝마다 100만 개를 다 계산해야 해서 느리다.
#    실무에서는 일부만 뽑아 쓰는 미니배치 방식(SGD)을 쓴다.
# ===========================================================================
```
